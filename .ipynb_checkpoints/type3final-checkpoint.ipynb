```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.lines import Line2D

plt.rcParams.update({
    "font.size": 13,
    "axes.labelsize": 16,
    "legend.fontsize": 11,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "axes.linewidth": 1.0,
    "lines.linewidth": 1.6,
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

col_beta0   = "#1f77b4"   # blu
col_betaneq = "#2ca02c"   # verde
col_wrong   = "#d62728"   # rosso
col_lcdm    = "black"

out = Path("/home/youri/AxiCLASS/outputpaper")

pk0_lcdm_file    = out / "lcdmz000_pk.dat"
pk0_beta0_file   = out / "EDE_z000_pk.dat"
pk0_betaneq_file = out / "EDE_rightz000_pk.dat"
pk0_wrong_file   = out / "EDE_wrongz000_pk.dat"

pk1000_lcdm_file    = out / "lcdmz100000_pk.dat"
pk1000_beta0_file   = out / "EDE_z100000_pk.dat"
pk1000_betaneq_file = out / "EDE_rightz100000_pk.dat"
pk1000_wrong_file   = out / "EDE_wrongz100000_pk.dat"

def load_pk(filename):
    data = np.loadtxt(filename, comments="#")
    k  = data[:, 0]
    pk = data[:, 1]
    return k, pk

def residual_interp(x_ref, x_model, y_model, y_ref):
    y_interp = np.interp(x_ref, x_model, y_model)
    y_ref = np.asarray(y_ref, dtype=float)
    out = np.full_like(y_ref, np.nan, dtype=float)
    mask = y_ref != 0.0
    out[mask] = (y_interp[mask] - y_ref[mask]) / y_ref[mask]
    return out

k0_lcdm, pk0_lcdm = load_pk(pk0_lcdm_file)
k0_beta0, pk0_beta0 = load_pk(pk0_beta0_file)
k0_betaneq, pk0_betaneq = load_pk(pk0_betaneq_file)
k0_wrong, pk0_wrong = load_pk(pk0_wrong_file)

pk0_res_beta0   = residual_interp(k0_lcdm, k0_beta0, pk0_beta0, pk0_lcdm)
pk0_res_betaneq = residual_interp(k0_lcdm, k0_betaneq, pk0_betaneq, pk0_lcdm)
pk0_res_wrong   = residual_interp(k0_lcdm, k0_wrong, pk0_wrong, pk0_lcdm)

k1000_lcdm, pk1000_lcdm = load_pk(pk1000_lcdm_file)
k1000_beta0, pk1000_beta0 = load_pk(pk1000_beta0_file)
k1000_betaneq, pk1000_betaneq = load_pk(pk1000_betaneq_file)
k1000_wrong, pk1000_wrong = load_pk(pk1000_wrong_file)

pk1000_res_beta0   = residual_interp(k1000_lcdm, k1000_beta0, pk1000_beta0, pk1000_lcdm)
pk1000_res_betaneq = residual_interp(k1000_lcdm, k1000_betaneq, pk1000_betaneq, pk1000_lcdm)
pk1000_res_wrong   = residual_interp(k1000_lcdm, k1000_wrong, pk1000_wrong, pk1000_lcdm)

fig, ax = plt.subplots(2, 1, figsize=(7, 8), sharex=True)

sigma_handles = [
    Line2D([0], [0], color=col_betaneq, lw=1.8,
           label=r'EDE with $\beta \neq 0$: $\sigma_8 = 0.79$'),
    Line2D([0], [0], color=col_beta0, lw=1.8,
           label=r'EDE with $\beta = 0$: $\sigma_8 = 0.86$')
]

# z = 0
line_pk0_beta0, = ax[0].semilogx(k0_lcdm, pk0_res_beta0, color=col_beta0,
                                 label=r'EDE with $\beta = 0$')
line_pk0_betaneq, = ax[0].semilogx(k0_lcdm, pk0_res_betaneq, color=col_betaneq,
                                   label=r'EDE with $\beta \neq 0$')
line_pk0_wrong, = ax[0].semilogx(k0_lcdm, pk0_res_wrong, color=col_wrong,
                                 label=r'EDE with $\beta \neq 0$ wrong')
line_pk0_lcdm, = ax[0].semilogx(k0_lcdm, np.zeros_like(k0_lcdm), '--', color=col_lcdm,
                                label=r'$\Lambda$CDM')

ax[0].axhline(0, color='gray', lw=0.8)
ax[0].set_xlim(1e-4, 1e-1)
ax[0].set_ylabel(r'$\Delta P(k)/P(k)$')
ax[0].text(0.95, 0.92, r'$z=0$', transform=ax[0].transAxes, ha='right', va='top')

leg1 = ax[0].legend(
    handles=[line_pk0_beta0, line_pk0_betaneq, line_pk0_wrong, line_pk0_lcdm],
    loc='upper left',
    frameon=True
)
leg2 = ax[0].legend(
    handles=sigma_handles,
    loc='lower left',
    frameon=True
)
ax[0].add_artist(leg1)

# z = 1000
line_pk1000_beta0, = ax[1].semilogx(k1000_lcdm, pk1000_res_beta0, color=col_beta0,
                                    label=r'EDE with $\beta = 0$')
line_pk1000_betaneq, = ax[1].semilogx(k1000_lcdm, pk1000_res_betaneq, color=col_betaneq,
                                      label=r'EDE with $\beta \neq 0$')
line_pk1000_wrong, = ax[1].semilogx(k1000_lcdm, pk1000_res_wrong, color=col_wrong,
                                    label=r'EDE with $\beta \neq 0$ wrong')
line_pk1000_lcdm, = ax[1].semilogx(k1000_lcdm, np.zeros_like(k1000_lcdm), '--', color=col_lcdm,
                                   label=r'$\Lambda$CDM')

ax[1].axhline(0, color='gray', lw=0.8)
ax[1].set_xlim(1e-4, 1e-1)
ax[1].set_xlabel(r'$k\,[h/\mathrm{Mpc}]$')
ax[1].set_ylabel(r'$\Delta P(k)/P(k)$')
ax[1].text(0.95, 0.92, r'$z=1000$', transform=ax[1].transAxes, ha='right', va='top')

leg3 = ax[1].legend(
    handles=[line_pk1000_beta0, line_pk1000_betaneq, line_pk1000_wrong, line_pk1000_lcdm],
    loc='upper left',
    frameon=True
)
leg4 = ax[1].legend(
    handles=sigma_handles,
    loc='lower left',
    frameon=True
)
#ax[1].add_artist(leg3)

plt.subplots_adjust(
    left=0.14,
    right=0.98,
    bottom=0.10,
    top=0.97,
    hspace=0.10
)

plt.savefig("pk_only_with_wrong.pdf", bbox_inches="tight")
plt.savefig("pk_only_with_wrong.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

plt.rcParams.update({
    "font.size": 13,
    "axes.labelsize": 16,
    "legend.fontsize": 11,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "axes.linewidth": 1.0,
    "lines.linewidth": 1.6,
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

col_beta0   = "#1f77b4"   # blu
col_betaneq = "#2ca02c"   # verde
col_wrong   = "#d62728"   # rosso
col_lcdm    = "black"

out = Path("/home/youri/AxiCLASS/outputpaper")

cl_lcdm_file    = out / "lcdmz000_cl.dat"
cl_beta0_file   = out / "EDE_z000_cl.dat"
cl_betaneq_file = out / "EDE_rightz000_cl.dat"
cl_wrong_file   = out / "EDE_wrongz000_cl.dat"

def load_cl(filename):
    data = np.loadtxt(filename, comments="#")
    ell = data[:, 0]
    tt  = data[:, 1]
    ee  = data[:, 2]
    return ell, tt, ee

def residual_interp(x_ref, x_model, y_model, y_ref):
    y_interp = np.interp(x_ref, x_model, y_model)
    y_ref = np.asarray(y_ref, dtype=float)
    out = np.full_like(y_ref, np.nan, dtype=float)
    mask = y_ref != 0.0
    out[mask] = (y_interp[mask] - y_ref[mask]) / y_ref[mask]
    return out

ell_lcdm, tt_lcdm, ee_lcdm = load_cl(cl_lcdm_file)
ell_beta0, tt_beta0, ee_beta0 = load_cl(cl_beta0_file)
ell_betaneq, tt_betaneq, ee_betaneq = load_cl(cl_betaneq_file)
ell_wrong, tt_wrong, ee_wrong = load_cl(cl_wrong_file)

tt_res_beta0   = residual_interp(ell_lcdm, ell_beta0, tt_beta0, tt_lcdm)
tt_res_betaneq = residual_interp(ell_lcdm, ell_betaneq, tt_betaneq, tt_lcdm)
tt_res_wrong   = residual_interp(ell_lcdm, ell_wrong, tt_wrong, tt_lcdm)

ee_res_beta0   = residual_interp(ell_lcdm, ell_beta0, ee_beta0, ee_lcdm)
ee_res_betaneq = residual_interp(ell_lcdm, ell_betaneq, ee_betaneq, ee_lcdm)
ee_res_wrong   = residual_interp(ell_lcdm, ell_wrong, ee_wrong, ee_lcdm)

fig, ax = plt.subplots(2, 1, figsize=(7.5, 8), sharex=True)

# TT
ax[0].plot(ell_lcdm, tt_res_beta0, color=col_beta0, label=r'EDE with $\beta = 0$')
ax[0].plot(ell_lcdm, tt_res_betaneq, color=col_betaneq, label=r'EDE with $\beta \neq 0$')
ax[0].plot(ell_lcdm, tt_res_wrong, color=col_wrong, label=r'EDE with $\beta \neq 0$ wrong')
ax[0].plot(ell_lcdm, np.zeros_like(ell_lcdm), '--', color=col_lcdm, label=r'$\Lambda$CDM')

ax[0].axhline(0, color='gray', lw=0.8)
ax[0].set_xlim(100, 2000)
ax[0].set_ylim(-0.1, 0.1)
ax[0].set_ylabel(r'$\Delta C_\ell^{TT}/C_\ell^{TT}$')
ax[0].legend(loc='upper center', ncol=2, frameon=True)

# EE
ax[1].plot(ell_lcdm, ee_res_beta0, color=col_beta0)
ax[1].plot(ell_lcdm, ee_res_betaneq, color=col_betaneq)
ax[1].plot(ell_lcdm, ee_res_wrong, color=col_wrong)
ax[1].plot(ell_lcdm, np.zeros_like(ell_lcdm), '--', color=col_lcdm)

ax[1].axhline(0, color='gray', lw=0.8)
ax[1].set_xlim(100, 2000)
ax[1].set_ylim(-0.2, 0.4)
ax[1].set_xlabel(r'$\ell$')
ax[1].set_ylabel(r'$\Delta C_\ell^{EE}/C_\ell^{EE}$')

plt.subplots_adjust(
    left=0.14,
    right=0.98,
    bottom=0.10,
    top=0.97,
    hspace=0.08
)

plt.savefig("cl_only_with_wrong.pdf", bbox_inches="tight")
plt.savefig("cl_only_with_wrong.png", dpi=300, bbox_inches="tight")
plt.show()
```
