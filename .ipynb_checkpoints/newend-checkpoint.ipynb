```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.lines import Line2D

plt.rcParams.update({
    "font.size": 13,
    "axes.labelsize": 16,
    "legend.fontsize": 11,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "axes.linewidth": 1.0,
    "lines.linewidth": 1.6,
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

col_beta0   = "#1f77b4"   # blu
col_betaneq = "#2ca02c"   # verde
col_wrong   = "#d62728"   # rosso
col_lcdm    = "black"

out = Path("/home/youri/AxiCLASS/outputpaper")

pk_lcdm_file    = out / "lcdmz000_pk.dat"
pk_beta0_file   = out / "EDE_z000_pk.dat"
pk_betaneq_file = out / "EDE_rightz000_pk.dat"
pk_wrong_file   = out / "EDE_wrongz000_pk.dat"

clphi_lcdm_file    = out / "lcdmz000_cl_lensed.dat"
clphi_beta0_file   = out / "EDE_z000_cl_lensed.dat"
clphi_betaneq_file = out / "EDE_rightz000_cl_lensed.dat"
clphi_wrong_file   = out / "EDE_wrongz000_cl_lensed.dat"

def load_pk(filename):
    data = np.loadtxt(filename, comments="#")
    return data[:, 0], data[:, 1]

def load_cl_phiphi(filename):
    data = np.loadtxt(filename, comments="#")
    return data[:, 0], data[:, 5]   # colonna 6 = phiphi

def residual_interp(x_ref, x_model, y_model, y_ref):
    y_interp = np.interp(x_ref, x_model, y_model)
    y_ref = np.asarray(y_ref, dtype=float)
    out = np.full_like(y_ref, np.nan, dtype=float)
    mask = y_ref != 0.0
    out[mask] = (y_interp[mask] - y_ref[mask]) / y_ref[mask]
    return out

# P(k)
k_lcdm, pk_lcdm = load_pk(pk_lcdm_file)
k_beta0, pk_beta0 = load_pk(pk_beta0_file)
k_betaneq, pk_betaneq = load_pk(pk_betaneq_file)
k_wrong, pk_wrong = load_pk(pk_wrong_file)

pk_res_beta0   = residual_interp(k_lcdm, k_beta0, pk_beta0, pk_lcdm)
pk_res_betaneq = residual_interp(k_lcdm, k_betaneq, pk_betaneq, pk_lcdm)
pk_res_wrong   = residual_interp(k_lcdm, k_wrong, pk_wrong, pk_lcdm)

# C_l^{phiphi}
ell_lcdm, phiphi_lcdm = load_cl_phiphi(clphi_lcdm_file)
ell_beta0, phiphi_beta0 = load_cl_phiphi(clphi_beta0_file)
ell_betaneq, phiphi_betaneq = load_cl_phiphi(clphi_betaneq_file)
ell_wrong, phiphi_wrong = load_cl_phiphi(clphi_wrong_file)

phiphi_res_beta0   = residual_interp(ell_lcdm, ell_beta0, phiphi_beta0, phiphi_lcdm)
phiphi_res_betaneq = residual_interp(ell_lcdm, ell_betaneq, phiphi_betaneq, phiphi_lcdm)
phiphi_res_wrong   = residual_interp(ell_lcdm, ell_wrong, phiphi_wrong, phiphi_lcdm)

fig, ax = plt.subplots(2, 1, figsize=(7.5, 8))

sigma_handles = [
    Line2D([0], [0], color=col_betaneq, lw=1.8,
           label=r'EDE with $\beta \neq 0$: $\sigma_8 = 0.79$'),
    Line2D([0], [0], color=col_beta0, lw=1.8,
           label=r'EDE with $\beta = 0$: $\sigma_8 = 0.86$')
]

# P(k)
line_pk_beta0, = ax[0].semilogx(k_lcdm, pk_res_beta0, color=col_beta0,
                                label=r'EDE with $\beta = 0$')
line_pk_betaneq, = ax[0].semilogx(k_lcdm, pk_res_betaneq, color=col_betaneq,
                                  label=r'EDE with $\beta \neq 0$')
line_pk_wrong, = ax[0].semilogx(k_lcdm, pk_res_wrong, color=col_wrong,
                                label=r'EDE with $\beta \neq 0$ wrong')
line_pk_lcdm, = ax[0].semilogx(k_lcdm, np.zeros_like(k_lcdm), '--', color=col_lcdm,
                               label=r'$\Lambda$CDM')

ax[0].axhline(0, color='gray', lw=0.8)
ax[0].set_xlim(1e-4, 1e-1)
ax[0].set_ylabel(r'$\Delta P(k)/P(k)$')

leg1 = ax[0].legend(
    handles=[line_pk_beta0, line_pk_betaneq, line_pk_wrong, line_pk_lcdm],
    loc='upper left',
    frameon=True
)
leg2 = ax[0].legend(
    handles=sigma_handles,
    loc='lower left',
    frameon=True
)
ax[0].add_artist(leg1)

# C_l^{phiphi}
ax[1].plot(ell_lcdm, phiphi_res_beta0, color=col_beta0,
           label=r'EDE with $\beta = 0$')
ax[1].plot(ell_lcdm, phiphi_res_betaneq, color=col_betaneq,
           label=r'EDE with $\beta \neq 0$')
ax[1].plot(ell_lcdm, phiphi_res_wrong, color=col_wrong,
           label=r'EDE with $\beta \neq 0$ wrong')
ax[1].plot(ell_lcdm, np.zeros_like(ell_lcdm), '--', color=col_lcdm,
           label=r'$\Lambda$CDM')

ax[1].axhline(0, color='gray', lw=0.8)
ax[1].set_xlim(100, 2000)
ax[1].set_xlabel(r'$\ell$')
ax[1].set_ylabel(r'$\Delta C_\ell^{\phi\phi}/C_\ell^{\phi\phi}$')
#ax[1].legend(loc='upper left', frameon=True)

plt.subplots_adjust(
    left=0.14,
    right=0.98,
    bottom=0.10,
    top=0.97,
    hspace=0.12
)

plt.savefig("pk_phiphi_wrong.pdf", bbox_inches="tight")
#plt.savefig("pk_phiphi_with_wrong.png", dpi=300, bbox_inches="tight")
plt.show()
```
