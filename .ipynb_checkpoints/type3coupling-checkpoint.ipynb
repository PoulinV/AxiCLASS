```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# =========================
# STILE
# =========================
plt.rcParams.update({
    "font.size": 13,
    "axes.labelsize": 14,
    "legend.fontsize": 11,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "figure.figsize": (12,10),
    "axes.linewidth": 1.2,
    "lines.linewidth": 2.2,
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

col_lcdm = "black"
col_b0   = "#1f77b4"
col_bn   = "#d62728"

# =========================
# PATH
# =========================
out = Path("/home/youri/AxiCLASS/output")

cl_lcdm_file = out / "explanatory00_cl.dat"
cl_b0_file   = out / "EDE_almostnoidm24_cl.dat"
cl_bn_file   = out / "EDE_almostnoidm22_cl.dat"

pk_lcdm_file = out / "explanatory00_pk.dat"
pk_b0_file   = out / "EDE_almostnoidm24_pk.dat"
pk_bn_file   = out / "EDE_almostnoidm22_pk.dat"

# =========================
# LETTURA
# =========================
def load_cl_tt(f):
    d = np.loadtxt(f, comments="#")
    return d[:,0], d[:,1]

def load_pk(f):
    d = np.loadtxt(f, comments="#")
    return d[:,0], d[:,1]

# =========================
# DATI
# =========================
ell_lcdm, tt_lcdm = load_cl_tt(cl_lcdm_file)
ell_b0,   tt_b0   = load_cl_tt(cl_b0_file)
ell_bn,   tt_bn   = load_cl_tt(cl_bn_file)

k_lcdm, pk_lcdm = load_pk(pk_lcdm_file)
k_b0,   pk_b0   = load_pk(pk_b0_file)
k_bn,   pk_bn   = load_pk(pk_bn_file)

# =========================
# INTERPOLAZIONE + RATIO
# =========================
tt_b0_i = np.interp(ell_lcdm, ell_b0, tt_b0)
tt_bn_i = np.interp(ell_lcdm, ell_bn, tt_bn)

pk_b0_i = np.interp(k_lcdm, k_b0, pk_b0)
pk_bn_i = np.interp(k_lcdm, k_bn, pk_bn)

tt_ratio_b0 = tt_b0_i / tt_lcdm
tt_ratio_bn = tt_bn_i / tt_lcdm

pk_ratio_b0 = pk_b0_i / pk_lcdm
pk_ratio_bn = pk_bn_i / pk_lcdm

# =========================
# FIGURA
# =========================
fig, ax = plt.subplots(2, 2)

# ---- CMB ----
ax[0,0].semilogx(ell_lcdm, tt_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
ax[0,0].semilogx(ell_b0, tt_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[0,0].semilogx(ell_bn, tt_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')

ax[0,0].set_xlabel(r'$\ell$')
ax[0,0].set_ylabel(r'$TT$')
ax[0,0].legend()
ax[0,0].grid(alpha=0.25)

# ---- CMB ratio ----
ax[0,1].semilogx(ell_lcdm, tt_ratio_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[0,1].semilogx(ell_lcdm, tt_ratio_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
ax[0,1].axhline(1, color='black', lw=1, label=r'$\Lambda$CDM')

ax[0,1].set_xlabel(r'$\ell$')
ax[0,1].set_ylabel(r'$TT / TT_{\Lambda\mathrm{CDM}}$')
ax[0,1].legend()
ax[0,1].grid(alpha=0.25)

# ---- PK ----
ax[1,0].loglog(k_lcdm, pk_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
ax[1,0].loglog(k_b0, pk_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[1,0].loglog(k_bn, pk_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')

ax[1,0].set_xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
ax[1,0].set_ylabel(r'$P(k)$')
ax[1,0].legend()
ax[1,0].grid(alpha=0.25, which='both')

# ---- PK ratio ----
ax[1,1].semilogx(k_lcdm, pk_ratio_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[1,1].semilogx(k_lcdm, pk_ratio_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
ax[1,1].axhline(1, color='black', lw=1, label=r'$\Lambda$CDM')

ax[1,1].set_xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
ax[1,1].set_ylabel(r'$P(k) / P(k)_{\Lambda\mathrm{CDM}}$')
ax[1,1].legend()
ax[1,1].grid(alpha=0.25)

plt.tight_layout()
plt.savefig("all.pdf", bbox_inches="tight")
plt.show()
```

```python
# =========================
# 1) CMB TT
# =========================
plt.figure(figsize=(6,5))

plt.semilogx(ell_lcdm, tt_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
plt.semilogx(ell_b0, tt_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
plt.semilogx(ell_bn, tt_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')

plt.xlabel(r'$\ell$')
plt.ylabel(r'$TT$')
plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.savefig("cmb_tt.pdf", bbox_inches="tight")
plt.show()


# =========================
# 2) CMB ratio
# =========================
plt.figure(figsize=(6,5))

plt.semilogx(ell_lcdm, tt_ratio_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
plt.semilogx(ell_lcdm, tt_ratio_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
plt.axhline(1, color='black', lw=1, label=r'$\Lambda$CDM')

plt.xlabel(r'$\ell$')
plt.ylabel(r'$TT / TT_{\Lambda\mathrm{CDM}}$')
plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.savefig("cmb_ratio.pdf", bbox_inches="tight")
plt.show()


# =========================
# 3) Matter power spectrum
# =========================
plt.figure(figsize=(6,5))

plt.loglog(k_lcdm, pk_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
plt.loglog(k_b0, pk_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
plt.loglog(k_bn, pk_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')

plt.xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$P(k)$')
plt.legend()
plt.grid(alpha=0.25, which='both')

plt.tight_layout()
plt.savefig("pk.pdf", bbox_inches="tight")
plt.show()


# =========================
# 4) Matter ratio
# =========================
plt.figure(figsize=(6,5))

plt.semilogx(k_lcdm, pk_ratio_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
plt.semilogx(k_lcdm, pk_ratio_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
plt.axhline(1, color='black', lw=1, label=r'$\Lambda$CDM')

plt.xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
plt.ylabel(r'$P(k) / P(k)_{\Lambda\mathrm{CDM}}$')
plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.savefig("pk_ratio.pdf", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# =========================
# STILE
# =========================
plt.rcParams.update({
    "font.size": 13,
    "axes.labelsize": 14,
    "legend.fontsize": 11,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "figure.figsize": (12,10),
    "axes.linewidth": 1.2,
    "lines.linewidth": 2.2,
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

col_lcdm  = "black"
col_b0    = "#1f77b4"
col_bn    = "#d62728"
col_wrong = "#2ca02c"

# =========================
# PATH
# =========================
out = Path("/home/youri/AxiCLASS/output")

cl_lcdm_file  = out / "explanatory00_cl.dat"
cl_b0_file    = out / "EDE_almostnoidm24_cl.dat"
cl_bn_file    = out / "EDE_almostnoidm22_cl.dat"
cl_wrong_file = out / "EDE_almostnoidm25_cl.dat"

pk_lcdm_file  = out / "explanatory00_pk.dat"
pk_b0_file    = out / "EDE_almostnoidm24_pk.dat"
pk_bn_file    = out / "EDE_almostnoidm22_pk.dat"
pk_wrong_file = out / "EDE_almostnoidm25_pk.dat"

# =========================
# LETTURA
# =========================
def load_cl_tt(f):
    d = np.loadtxt(f, comments="#")
    return d[:,0], d[:,1]

def load_pk(f):
    d = np.loadtxt(f, comments="#")
    return d[:,0], d[:,1]

# =========================
# DATI
# =========================
ell_lcdm,  tt_lcdm  = load_cl_tt(cl_lcdm_file)
ell_b0,    tt_b0    = load_cl_tt(cl_b0_file)
ell_bn,    tt_bn    = load_cl_tt(cl_bn_file)
ell_wrong, tt_wrong = load_cl_tt(cl_wrong_file)

k_lcdm,  pk_lcdm  = load_pk(pk_lcdm_file)
k_b0,    pk_b0    = load_pk(pk_b0_file)
k_bn,    pk_bn    = load_pk(pk_bn_file)
k_wrong, pk_wrong = load_pk(pk_wrong_file)

# =========================
# INTERPOLAZIONE + RATIO
# =========================
tt_b0_i    = np.interp(ell_lcdm, ell_b0, tt_b0)
tt_bn_i    = np.interp(ell_lcdm, ell_bn, tt_bn)
tt_wrong_i = np.interp(ell_lcdm, ell_wrong, tt_wrong)

pk_b0_i    = np.interp(k_lcdm, k_b0, pk_b0)
pk_bn_i    = np.interp(k_lcdm, k_bn, pk_bn)
pk_wrong_i = np.interp(k_lcdm, k_wrong, pk_wrong)

tt_ratio_b0    = tt_b0_i / tt_lcdm
tt_ratio_bn    = tt_bn_i / tt_lcdm
tt_ratio_wrong = tt_wrong_i / tt_lcdm

pk_ratio_b0    = pk_b0_i / pk_lcdm
pk_ratio_bn    = pk_bn_i / pk_lcdm
pk_ratio_wrong = pk_wrong_i / pk_lcdm

# =========================
# FIGURA
# =========================
fig, ax = plt.subplots(2, 2)

# ---- CMB ----
ax[0,0].semilogx(ell_lcdm, tt_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
ax[0,0].semilogx(ell_b0, tt_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[0,0].semilogx(ell_bn, tt_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
ax[0,0].semilogx(ell_wrong, tt_wrong, ':', color=col_wrong, label=r"EDE $\beta\neq 0$ wrong")

ax[0,0].set_xlabel(r'$\ell$')
ax[0,0].set_ylabel(r'$TT$')
ax[0,0].legend()
ax[0,0].grid(alpha=0.25)

# ---- CMB ratio ----
ax[0,1].semilogx(ell_lcdm, tt_ratio_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[0,1].semilogx(ell_lcdm, tt_ratio_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
ax[0,1].semilogx(ell_lcdm, tt_ratio_wrong, ':', color=col_wrong, label=r"EDE $\beta\neq 0$ wrong")
ax[0,1].axhline(1, color='black', lw=1, label=r'$\Lambda$CDM')

ax[0,1].set_xlabel(r'$\ell$')
ax[0,1].set_ylabel(r'$TT / TT_{\Lambda\mathrm{CDM}}$')
ax[0,1].legend()
ax[0,1].grid(alpha=0.25)

# ---- PK ----
ax[1,0].loglog(k_lcdm, pk_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
ax[1,0].loglog(k_b0, pk_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[1,0].loglog(k_bn, pk_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
ax[1,0].loglog(k_wrong, pk_wrong, ':', color=col_wrong, label=r"EDE $\beta\neq 0$ wrong")

ax[1,0].set_xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
ax[1,0].set_ylabel(r'$P(k)$')
ax[1,0].legend()
ax[1,0].grid(alpha=0.25, which='both')

# ---- PK ratio ----
ax[1,1].semilogx(k_lcdm, pk_ratio_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[1,1].semilogx(k_lcdm, pk_ratio_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')
ax[1,1].semilogx(k_lcdm, pk_ratio_wrong, ':', color=col_wrong, label=r"EDE $\beta\neq 0$ wrong")
ax[1,1].axhline(1, color='black', lw=1, label=r'$\Lambda$CDM')

ax[1,1].set_xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
ax[1,1].set_ylabel(r'$P(k) / P(k)_{\Lambda\mathrm{CDM}}$')
ax[1,1].legend()
ax[1,1].grid(alpha=0.25)

plt.tight_layout()
plt.savefig("all2.pdf", bbox_inches="tight")
plt.show()
```
