```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# =========================
# STILE "PAPER"
# =========================
plt.rcParams.update({
    "font.size": 13,
    "axes.labelsize": 14,
    "axes.titlesize": 15,
    "legend.fontsize": 11,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "figure.figsize": (12, 10),
    "axes.linewidth": 1.2,
    "lines.linewidth": 2.2,
    "text.usetex": False,   # evita errore latex
    "font.family": "serif",
    "mathtext.fontset": "cm"
})

# =========================
# COLORI
# =========================
col_lcdm = "black"
col_b0   = "#1f77b4"   # blu
col_bn   = "#d62728"   # rosso

# =========================
# PATH
# =========================
out = Path("/home/youri/AxiCLASS/output")

# file CMB
cl_lcdm_file = out / "explanatory00_cl.dat"
cl_b0_file   = out / "EDE_almostnoidm24_cl.dat"   # beta = 0
cl_bn_file   = out / "EDE_almostnoidm22_cl.dat"   # beta != 0

# file matter power spectrum
pk_lcdm_file = out / "explanatory00_pk.dat"
pk_b0_file   = out / "EDE_almostnoidm24_pk.dat"   # beta = 0
pk_bn_file   = out / "EDE_almostnoidm22_pk.dat"   # beta != 0

# =========================
# FUNZIONI DI LETTURA
# =========================
def load_cl_tt(filename):
    data = np.loadtxt(filename, comments="#")
    ell = data[:, 0]
    tt  = data[:, 1]
    return ell, tt

def load_pk(filename):
    data = np.loadtxt(filename, comments="#")
    k  = data[:, 0]
    pk = data[:, 1]
    return k, pk

# =========================
# CARICAMENTO DATI
# =========================
ell_lcdm, tt_lcdm = load_cl_tt(cl_lcdm_file)
ell_b0,   tt_b0   = load_cl_tt(cl_b0_file)
ell_bn,   tt_bn   = load_cl_tt(cl_bn_file)

k_lcdm, pk_lcdm = load_pk(pk_lcdm_file)
k_b0,   pk_b0   = load_pk(pk_b0_file)
k_bn,   pk_bn   = load_pk(pk_bn_file)

# =========================
# INTERPOLAZIONE SU LCDM
# =========================
tt_b0_interp = np.interp(ell_lcdm, ell_b0, tt_b0)
tt_bn_interp = np.interp(ell_lcdm, ell_bn, tt_bn)

pk_b0_interp = np.interp(k_lcdm, k_b0, pk_b0)
pk_bn_interp = np.interp(k_lcdm, k_bn, pk_bn)

# =========================
# RATIO RISPETTO A LCDM
# =========================
tt_ratio_b0 = tt_b0_interp / tt_lcdm
tt_ratio_bn = tt_bn_interp / tt_lcdm

pk_ratio_b0 = pk_b0_interp / pk_lcdm
pk_ratio_bn = pk_bn_interp / pk_lcdm

# =========================
# FIGURA
# =========================
fig, ax = plt.subplots(2, 2)

# ---- CMB TT ----
ax[0, 0].semilogx(ell_lcdm, tt_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
ax[0, 0].semilogx(ell_b0, tt_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[0, 0].semilogx(ell_bn, tt_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')

ax[0, 0].set_xlabel(r'$\ell$')
ax[0, 0].set_ylabel(r'$TT$')
ax[0, 0].set_title('CMB TT Spectrum')
ax[0, 0].legend()
ax[0, 0].grid(alpha=0.25)

# ---- CMB ratio ----
ax[0, 1].semilogx(ell_lcdm, tt_ratio_b0, '--', color=col_b0,
                  label=r'EDE $\beta=0$ / $\Lambda$CDM')
ax[0, 1].semilogx(ell_lcdm, tt_ratio_bn, '-.', color=col_bn,
                  label=r'EDE $\beta\neq 0$ / $\Lambda$CDM')

ax[0, 1].axhline(1.0, color='black', lw=1)
ax[0, 1].set_xlabel(r'$\ell$')
ax[0, 1].set_ylabel('Ratio')
ax[0, 1].set_title('CMB / LCDM')
ax[0, 1].legend()
ax[0, 1].grid(alpha=0.25)

# ---- Matter power spectrum ----
ax[1, 0].loglog(k_lcdm, pk_lcdm, color=col_lcdm, label=r'$\Lambda$CDM')
ax[1, 0].loglog(k_b0, pk_b0, '--', color=col_b0, label=r'EDE $\beta=0$')
ax[1, 0].loglog(k_bn, pk_bn, '-.', color=col_bn, label=r'EDE $\beta\neq 0$')

ax[1, 0].set_xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
ax[1, 0].set_ylabel(r'$P(k)$')
ax[1, 0].set_title('Matter Power Spectrum')
ax[1, 0].legend()
ax[1, 0].grid(alpha=0.25, which='both')

# ---- Matter ratio ----
ax[1, 1].semilogx(k_lcdm, pk_ratio_b0, '--', color=col_b0,
                  label=r'EDE $\beta=0$ / $\Lambda$CDM')
ax[1, 1].semilogx(k_lcdm, pk_ratio_bn, '-.', color=col_bn,
                  label=r'EDE $\beta\neq 0$ / $\Lambda$CDM')

ax[1, 1].axhline(1.0, color='black', lw=1)
ax[1, 1].set_xlabel(r'$k\,[\mathrm{Mpc}^{-1}]$')
ax[1, 1].set_ylabel('Ratio')
ax[1, 1].set_title(r'$P(k)$ / LCDM')
ax[1, 1].legend()
ax[1, 1].grid(alpha=0.25)

plt.tight_layout()

# =========================
# SALVATAGGIO
# =========================
plt.savefig("cosmo_plot.pdf", bbox_inches="tight")
plt.savefig("cosmo_plot.png", dpi=300)

plt.show()
```
